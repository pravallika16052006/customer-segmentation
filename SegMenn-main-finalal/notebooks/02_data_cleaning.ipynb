```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
print("Education values:")
print(df["Education"].value_counts())

print("\nMarital Status values:")
print(df["Marital_Status"].value_counts())
```

```python
df["Marital_Status"] = df["Marital_Status"].replace({
    "Alone": "Single",
    "Absurd": "Single",
    "YOLO": "Single"
})
```

```python
df["Marital_Status"].value_counts()
```

```python
print("Minimum Year of Birth:", df["Year_Birth"].min())
print("Maximum Year of Birth:", df["Year_Birth"].max())
```

```python
print("Invalid Year_Birth values:", (df["Year_Birth"] < 1940).sum())
```

```python
df[df["Year_Birth"] < 1940][["ID", "Year_Birth", "Income"]]
```

```python
df = df[df["Year_Birth"] >= 1940].copy()
```

```python
print("Rows after removing invalid birth years:", len(df))
print("Invalid birth years remaining:", (df["Year_Birth"] < 1940).sum())
```

```python
df.duplicated().sum()
```

```python
df["Education"] = df["Education"].replace({
    "2n Cycle": "2nd Cycle"
})
```

```python
df["Education"].value_counts()
```

```python
print("Income statistics:")
print(df["Income"].describe())

print("\nIncome outliers:")
Q1 = df["Income"].quantile(0.25)
Q3 = df["Income"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = df[(df["Income"] < lower_bound) | (df["Income"] > upper_bound)]

print("Number of income outliers:", len(outliers))
```

```python
print("Income statistics:")
print(df["Income"].describe())

print("\nIncome outliers:")
Q1 = df["Income"].quantile(0.25)
Q3 = df["Income"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = df[(df["Income"] < lower_bound) | (df["Income"] > upper_bound)]

print("Number of income outliers:", len(outliers))
```

```python
df["Income"] = df["Income"].clip(upper=upper_bound)

print("Maximum Income after capping:", df["Income"].max())
```

```python
print("Income outliers after capping:",
      ((df["Income"] < lower_bound) | (df["Income"] > upper_bound)).sum())
```

```python
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    errors="coerce",
    dayfirst=True
)

print("Invalid dates:", df["Dt_Customer"].isnull().sum())
```

```python
print(df["Dt_Customer"].min())
print(df["Dt_Customer"].max())
```

```python
print("Response values:")
print(df["Response"].value_counts())
```
